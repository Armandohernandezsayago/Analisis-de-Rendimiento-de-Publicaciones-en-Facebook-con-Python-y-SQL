**EXPLORACIÓN DE DATOS**

```python

# 1. IMPORTAR LIBRERÍAS

import pandas as pd
import numpy as np




# 2. CARGAR ARCHIVO CSV

ruta_archivo = "Datos Crudos Facebook 10 de Marzo.csv"

df = pd.read_csv(ruta_archivo)

print("Archivo cargado correctamente.")




# 3. DIMENSIONES DEL DATASET

print("Dimensiones del dataset:")
print(df.shape)



# 4. NOMBRES DE COLUMNAS

print("Columnas del dataset:\n")

for i, columna in enumerate(df.columns, start=1):
    print(f"{i}. {columna}")
```

```python

# 5. INFORMACIÓN GENERAL DEL DATASET

print("Información general:\n")

df.info()



# 6. PRIMERAS FILAS DEL DATASET

print("Primeras filas:\n")

display(df.head())



# 7. VALORES NULOS POR COLUMNA

nulos = df.isnull().sum().sort_values(ascending=False)

print("Valores nulos por columna:\n")

display(nulos)




# 8. PORCENTAJE DE VALORES NULOS

porcentaje_nulos = (
    df.isnull().mean() * 100
).sort_values(ascending=False)

print("Porcentaje de valores nulos:\n")

display(porcentaje_nulos)
```

```python

# 9. CANTIDAD DE CEROS POR COLUMNA

ceros = (
    (df == 0)
    .sum(numeric_only=True)
    .sort_values(ascending=False)
)

print("Cantidad de ceros por columna:\n")

display(ceros)




# 10. PORCENTAJE DE CEROS POR COLUMNA

porcentaje_ceros = (
    (df == 0)
    .mean(numeric_only=True) * 100
).sort_values(ascending=False)

print("Porcentaje de ceros por columna:\n")

display(porcentaje_ceros)




# 11. VALORES ÚNICOS POR COLUMNA

valores_unicos = (
    df.nunique()
    .sort_values()
)

print("Cantidad de valores únicos por columna:\n")

display(valores_unicos)



# 12. ESTADÍSTICAS DESCRIPTIVAS

print("Estadísticas descriptivas:\n")

display(df.describe())
```

```python

# 13. DETECTAR FILAS DUPLICADAS

duplicados = df.duplicated().sum()

print(f"Cantidad de filas duplicadas: {duplicados}")




# 14. BUSCAR COLUMNAS RELACIONADAS CON PAÍSES

columnas_paises = [
    col for col in df.columns
    if "país" in col.lower()
]

print("Columnas relacionadas con países:\n")

for col in columnas_paises:
    print(col)




# 15. BUSCAR COLUMNAS RELACIONADAS CON VIDEOS

columnas_video = [
    col for col in df.columns
    if "video" in col.lower()
]

print("Columnas relacionadas con video:\n")

for col in columnas_video:
    print(col)




# 16. BUSCAR COLUMNAS RELACIONADAS CON REACCIONES

columnas_reacciones = [
    col for col in df.columns
    if "reacción" in col.lower()
    or "like" in col.lower()
    or "coment" in col.lower()
    or "compart" in col.lower()
]

print("Columnas relacionadas con interacción:\n")

for col in columnas_reacciones:
    print(col)
```
